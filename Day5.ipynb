```python
import pandas as pd
import numpy as np
```

```python
data ={
    "name": ["max","snappy","slappy","aki","abhi"],
    "age": [25,30,np.nan,49,70,],
    "city": ["pune","pune","pune","tamilnadu","karnataka"]

}
print(data)
df = pd.DataFrame(data)
df
df.info()
df.head()
```

```python
print("mean of the age:")
print(df["age"].mean())
print("median of the age:")
print(df["age"].median())
print("mode of the age:")
print(df["age"].mode)
```

```python
new_data ={
    "name": ["max","snappy","slappy","aki","abhi"],
    "age": [25,30,np.nan,49,70,],
    "salary": [1,2,3,4,5]

}
print(new_data)
df = pd.DataFrame(new_data)
df
df.info()
df.head()
```

```python
df[["age","salary"]]= df[["age","salary"]].fillna(df[["age","salary"]]).mean()
df
```
